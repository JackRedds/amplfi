```python
import os
import sys
```

```python
def get_file_names(data_dir):
    """List of background files used for both training and validation"""
    background_dir = data_dir
    fnames = sorted((background_dir.glob("*.hdf5")))

    return fnames
```

```python
from pathlib import Path
data_dir = Path("/projects/bcse/jredepenning/gwak2_background")
fnames = get_file_names(data_dir)
```

```python
from torch.utils.data import random_split

dataset_size = len(fnames)
train_size = int(0.7 * dataset_size) + 1
test_size   = int(0.3 * dataset_size)
```

```python
train_set, test_set = random_split(
    fnames, [train_size, test_size]
)
```

```python
import shutil

test_destination = Path("/projects/bcse/jredepenning/gwak2_background/test/background/")
train_destination = Path("/projects/bcse/jredepenning/gwak2_background/test/background/")
```

```python
for i in train_set.indices:
    shutil.move(fnames[i], train_destination)
```

```python
for i in test_set.indices:
    shutil.move(fnames[i], test_destination)
```
